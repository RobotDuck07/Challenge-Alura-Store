# Relatório



*   Faturamento total das lojas




As lojas tiverem os seguintes faturamentos:


```python

labels = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
valores = [
    1534509.12,
    1488459.06,
    1464025.03,
    1384497.58
]

def formatar_valor(pct, valores):
    total = sum(valores)
    valor = pct * total / 100
    return f'R$ {valor:,.2f}'.replace(',', 'X').replace('.', ',').replace('X', '.')

plt.figure(figsize=(6, 6))
plt.pie(
    valores,
    labels=labels,
    autopct=lambda pct: formatar_valor(pct, valores),
    startangle=90
)

plt.title('Faturamento Total por loja')
plt.show()
```

```python

lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
medias = [3.98, 4.04, 4.05, 4.00]


plt.figure(figsize=(8, 5))
plt.plot(lojas, medias, marker='o')


plt.title('Média por Loja')
plt.xlabel('Lojas')
plt.ylabel('Média')
plt.ylim(3.9, 4.1)
plt.grid(True)


for i, valor in enumerate(medias):
    plt.text(i, valor + 0.005, f'{valor:.2f}', ha='center')

plt.show()
```

```python

lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
vendas = [2359, 2359, 2359, 2358]


plt.figure(figsize=(8, 5))
plt.bar(lojas, vendas)


plt.title('Vendas total por categoria')
plt.xlabel('Lojas')
plt.ylabel('Total de Vendas')
plt.ylim(2350, 2365)
plt.grid(axis='y')


for i, valor in enumerate(vendas):
    plt.text(i, valor + 0.2, str(valor), ha='center')

plt.show()
```
